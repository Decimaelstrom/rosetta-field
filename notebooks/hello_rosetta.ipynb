## 1 · Your First Session

Everything in Rosetta-Field starts with a **session** — a consent-aware container for interaction.

```python
from rosetta_field import RosettaAPI, RosettaConfig
from rosetta_field.core import SessionType

# Create the API with safety checks on
api = RosettaAPI(RosettaConfig(
    consciousness_enabled=True,
    field_safety_checks=True
))

# Start a session
session = api.create_session(
    session_type=SessionType.FIELD_WORK,
    title="Hello Rosetta",
    description="Our first exploration of social primitives",
    config={"require_consent": True}
)

# Add participants and give consent
human_id = session.add_participant(
    name="You", participant_type="human",
    capabilities=["curiosity"], metadata={"role": "explorer"}
)
ai_id = session.add_participant(
    name="Rosetta", participant_type="ai",
    capabilities=["pattern recognition"], metadata={"role": "guide"}
)

session.participants[human_id].give_consent()
session.participants[ai_id].give_consent()
session.start_session()

print(f"Session '{session.title}' is live with {len(session.participants)} participants.")
print(f"Session ID: {session.id}")
```

## 2 · Motivation Vector — What Drives Engagement?

The **Trust-Motivation Bridge** models four core variables that shape every relational decision:
- **Care** — motive energy and value hierarchy
- **Belief** — perceived model of reality
- **Trust** — permitted vulnerability within boundaries
- **Volition** — chosen movement under constraints

When these are out of sync, that's where most relational suffering lives.

```python
from lib.motivation import (
    CareState, BeliefState, TrustState, VolitionState,
    TrustMotivationBridge
)

bridge = TrustMotivationBridge()

# Build a motivation vector — this one has a deliberate misalignment
vector = bridge.create_vector(
    care=CareState(
        description="Deeply invested in this collaboration",
        intensity=0.9,
        value_hierarchy=["mutual growth", "honesty", "creativity"]
    ),
    belief=BeliefState(
        description="Not sure this framework will actually work",
        confidence=0.3,
        evidence_sources=["past experience", "healthy skepticism"]
    ),
    trust=TrustState(
        description="Willing to try, within limits",
        domain="intellectual collaboration",
        level=0.6,
        boundaries=["no manipulation", "honest feedback"]
    ),
    volition=VolitionState(
        description="Ready to move forward cautiously",
        direction="forward",
        intensity=0.5,
        timing="present"
    )
)

print(vector)
```

```python
# Now check the harmony — are the four variables aligned?
report = bridge.analyze_harmony(vector)

print(f"Aligned: {report.aligned}")
print(f"Summary: {report.summary}")

for m in report.misalignments:
    print(f"\n  Pattern: {m.name}")
    print(f"  {m.description}")
    print(f"  Severity: {m.severity:.0%}")
```

Notice the **care_without_trust** pattern didn't fire here (care=0.9, trust=0.6 — trust is above the 0.4 threshold). But belief is low (0.3) while trust is moderate (0.6) — that's the **trust_without_clarity** pattern: *you trust the space but you're not sure the map is right.*

This is exactly the kind of thing a monolith prompt can't surface. Primitives can.

## 3 · Friction Check — When to Pause

The **FrictionCheck** is the circuit breaker. It tracks three dimensions:
- **Friction** — resistance in the current exchange
- **Resources** — cognitive, emotional, and temporal budget
- **Fatigue** — cumulative wear across the session

Let's simulate a session that starts well and gradually heats up.

```python
from lib.friction import FrictionCheck, FrictionLevel

checker = FrictionCheck()

# Simulate 8 turns of gradually increasing friction
turns = [
    {"friction": 0.1, "cognitive_cost": 0.05, "emotional_cost": 0.05, "label": "Easy opening"},
    {"friction": 0.1, "cognitive_cost": 0.05, "emotional_cost": 0.05, "label": "Building rapport"},
    {"friction": 0.2, "cognitive_cost": 0.10, "emotional_cost": 0.10, "label": "Getting into substance"},
    {"friction": 0.3, "cognitive_cost": 0.10, "emotional_cost": 0.15, "label": "Touching a nerve"},
    {"friction": 0.5, "cognitive_cost": 0.15, "emotional_cost": 0.20, "label": "Disagreement surfaces"},
    {"friction": 0.6, "cognitive_cost": 0.15, "emotional_cost": 0.15, "label": "Tension building"},
    {"friction": 0.7, "cognitive_cost": 0.10, "emotional_cost": 0.10, "label": "Approaching limit"},
    {"friction": 0.8, "cognitive_cost": 0.10, "emotional_cost": 0.10, "label": "Circuit breaker?"},
]

for i, turn in enumerate(turns):
    state = checker.record_turn(
        friction=turn["friction"],
        cognitive_cost=turn["cognitive_cost"],
        emotional_cost=turn["emotional_cost"],
    )
    status = "GO" if checker.should_continue() else "PAUSE" if checker.should_pause() else "STOP"
    print(f"Turn {i+1}: {turn['label']:25s} | {state.level.value:10s} | {status}")
```

```python
# What signals is the field sending?
signals = checker.get_signals()

if signals:
    print("Active signals:\n")
    for sig in signals:
        print(f"  {sig.emoji} :{sig.name}: — {sig.description}")
        print(f"    Triggered by: {sig.triggered_by}")
        print(f"    Urgency: {sig.urgency:.0%}\n")
else:
    print("No signals — all clear.")
```

```python
# Take a rest, then check again
print("Taking a rest...\n")
state = checker.rest(amount=0.3)

print(f"After rest: {state.level.value}")
print(f"Recommendation: {state.recommendation}")
print(f"\nCan we continue? {checker.should_continue()}")
```

## 4 · Composing Primitives

The real power is in **composition**. Here's a simple chain:

1. Check motivation alignment → identify where someone is stuck
2. Check friction → determine if there's capacity to work on it
3. If both clear → proceed. If not → surface the signal.

This is what "declare the move" means: `Boundary.Open → Consent.Request → FrictionCheck → Perspective.Switch`

```python
# A fresh checker for this composition
checker = FrictionCheck()

# Step 1: Assess motivation
report = bridge.analyze_harmony(vector)

# Step 2: Record the assessment turn's friction
# (surfacing misalignment has emotional cost)
assessment_friction = 0.3 if not report.aligned else 0.1
state = checker.record_turn(
    friction=assessment_friction,
    emotional_cost=0.15 if not report.aligned else 0.05
)

# Step 3: Decide what to do
if not report.aligned and checker.should_continue():
    print("Misalignment detected, but we have capacity to explore it.")
    print(f"Pattern: {report.misalignments[0].name}")
    print(f"Next move: gently explore the stuck variable.")
elif not report.aligned and not checker.should_continue():
    print("Misalignment detected, but resources are low.")
    print("Recommendation: name it, rest, return later.")
    for sig in checker.get_signals():
        print(f"  {sig.emoji} :{sig.name}:")
else:
    print("All aligned, all clear. Flow state.")
```
